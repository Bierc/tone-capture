```python
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
import os
```

```python
# Carregar o áudio
y, sr = librosa.load("../data/Chorus/Chorus/Bridge/1-0.wav", sr=None)

# Aplicar STFT
D = librosa.stft(y, n_fft=2048, hop_length=512)

# Separar magnitude e fase
magnitude, phase = np.abs(D), np.angle(D)

# Salvar os dados em arquivos
np.save("../teste/spectrogram_magnitude.npy", magnitude)
np.save("../teste/spectrogram_phase.npy", phase)

# Salvar o espectrograma como imagem (opcional)
S_db = librosa.amplitude_to_db(magnitude, ref=np.max)
plt.figure(figsize=(10, 6))
librosa.display.specshow(S_db, sr=sr, hop_length=512, x_axis='time', y_axis='log')
plt.colorbar(format="%+2.0f dB")
plt.title("Espectrograma")
plt.savefig("spectrogram_image.png")
plt.close()

print("Espectrograma salvo como arquivos .npy e imagem.")
```

```python
import soundfile as sf

# Carregar os dados do espectrograma
magnitude = np.load("../teste/spectrogram_magnitude.npy")
phase = np.load("../teste/spectrogram_phase.npy")

# Reconstruir o espectrograma complexo
D_reconstructed = magnitude * np.exp(1j * phase)

# Aplicar ISTFT para reconstruir o áudio
y_reconstructed = librosa.istft(D_reconstructed, hop_length=512)

# Salvar o áudio reconstruído
sf.write("../teste/reconstructed_audio.wav", y_reconstructed, sr)
print("Áudio reconstruído salvo como 'reconstructed_audio.wav'")
```

```python
# Carregar os dados de magnitude
magnitude = np.load("../teste/spectrogram_magnitude.npy")

# Estimar a fase usando o método de Griffin-Lim
y_reconstructed = librosa.griffinlim(magnitude, hop_length=512, n_iter=32)

# Salvar o áudio reconstruído
sf.write("../teste/reconstructed_audio_griffinlim.wav", y_reconstructed, sr)
print("Áudio reconstruído com Griffin-Lim salvo como 'reconstructed_audio_griffinlim.wav'")
```
