```python
import os
import librosa
import soundfile as sf
```

```python
def normalize_audio(input_path, output_path, sample_rate=16000, duration=None):
    """
    Normaliza os áudios para mono, taxa de amostragem fixa e duração uniforme.

    Args:
        input_path (str): Caminho para o dataset de entrada.
        output_path (str): Caminho para salvar os áudios processados.
        sample_rate (int): Taxa de amostragem desejada (ex.: 16000).
        duration (float): Duração fixa dos áudios em segundos (opcional).
    """
    os.makedirs(output_path, exist_ok=True)

    for folder in os.listdir(input_path):
        folder_path = os.path.join(input_path, folder)

        # Verifica se é um diretório
        if not os.path.isdir(folder_path):
            print(f"Ignorando {folder_path}, não é um diretório.")
            continue

        output_folder_path = os.path.join(output_path, folder)
        os.makedirs(output_folder_path, exist_ok=True)

        for file in os.listdir(folder_path):
            file_path = os.path.join(folder_path, file)

            # Processa apenas arquivos .wav
            if not file.endswith('.wav'):
                print(f"Ignorando {file_path}, não é um arquivo .wav.")
                continue

            try:
                # Carregar áudio
                y, sr = librosa.load(file_path, sr=sample_rate, mono=True)
                
                # Ajustar duração (adicionar ou cortar)
                if duration:
                    max_samples = int(sample_rate * duration)
                    y = librosa.util.fix_length(y, size=max_samples)

                # Salvar áudio processado
                output_file_path = os.path.join(output_folder_path, file)
                sf.write(output_file_path, y, sample_rate)
                print(f"Processado com sucesso: {output_file_path}")
            except Exception as e:
                print(f"Erro ao processar {file_path}: {e}")
```

```python
# Caminho dos dados de entrada e saída
input_path = "../data/Hall-Reverb/Hall-Reverb"  # Substituir pelo caminho do seu dataset
output_path = "../preprocessed_data/Hall-Reverb"

# Normalizar os áudios com taxa de 16kHz e duração fixa de 5 segundos
normalize_audio(input_path, output_path, sample_rate=16000, duration=5)
```
