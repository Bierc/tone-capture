```python
import os
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
```

```python
def generate_spectrograms(input_path, output_path, n_mels=128, fmax=8000):
    """
    Gera espectrogramas Mel a partir de arquivos .wav e salva como imagens PNG, mantendo a estrutura de subdiretórios.

    Args:
        input_path (str): Caminho para os áudios preprocessados.
        output_path (str): Caminho para salvar os espectrogramas.
        n_mels (int): Número de bandas Mel.
        fmax (int): Frequência máxima para o espectrograma.
    """
    for root, _, files in os.walk(input_path):  # Percorrer subdiretórios
        relative_path = os.path.relpath(root, input_path)
        output_dir = os.path.join(output_path, relative_path)
        os.makedirs(output_dir, exist_ok=True)  # Cria a estrutura correspondente no diretório de saída

        for file in files:
            if file.endswith('.wav'):
                try:
                    file_path = os.path.join(root, file)

                    # Carregar áudio
                    y, sr = librosa.load(file_path, sr=None)
                    
                    # Gerar espectrograma Mel
                    mel_spectrogram = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, fmax=fmax)
                    mel_db = librosa.power_to_db(mel_spectrogram, ref=np.max)

                    # Salvar como imagem PNG
                    plt.figure(figsize=(5, 5))
                    librosa.display.specshow(mel_db, sr=sr, fmax=fmax, x_axis='time', y_axis='mel', cmap='viridis')
                    plt.axis('off')
                    plt.tight_layout()
                    
                    output_file = os.path.join(output_dir, file.replace('.wav', '.png'))
                    plt.savefig(output_file, bbox_inches='tight', pad_inches=0)
                    plt.close()
                    print(f"Espectrograma salvo: {output_file}")
                except Exception as e:
                    print(f"Erro ao gerar espectrograma para {file_path}: {e}")
```

```python
# Caminhos de entrada e saída
input_path = "../dataset"  # Dataset preprocessado
output_path = "../spectrograms"  # Caminho para os espectrogramas

# Gerar espectrogramas
generate_spectrograms(input_path, output_path, n_mels=128, fmax=16384)
```
