```python
import os
import shutil
from sklearn.model_selection import train_test_split
```

```python
# Caminhos dos dados originais
input_dir = "../preprocessed_data/Chorus"
target_dir = "../preprocessed_data/Hall-Reverb"
```

```python
# Caminhos de saída
train_input_dir = "../dataset/train/input"
train_target_dir = "../dataset/train/target"
val_input_dir = "../dataset/val/input"
val_target_dir = "../dataset/val/target"
```

```python
def collect_files_recursively(root_dir):
    """
    Coleta todos os arquivos .wav em um diretório, incluindo subpastas.

    Args:
        root_dir (str): Caminho do diretório raiz.

    Returns:
        list: Lista de caminhos absolutos dos arquivos .wav.
    """
    file_list = []
    for root, _, files in os.walk(root_dir):
        for file in files:
            if file.endswith(".wav"):
                file_list.append(os.path.join(root, file))
    return file_list
```

```python
# Criar pastas de saída, se não existirem
os.makedirs(train_input_dir, exist_ok=True)
os.makedirs(train_target_dir, exist_ok=True)
os.makedirs(val_input_dir, exist_ok=True)
os.makedirs(val_target_dir, exist_ok=True)

# Coletar arquivos recursivamente
input_files = sorted(collect_files_recursively(input_dir))
target_files = sorted(collect_files_recursively(target_dir))

# Diagnóstico: imprimir arquivos encontrados
print(f"Arquivos de entrada encontrados: {len(input_files)}")
print(f"Arquivos de saída encontrados: {len(target_files)}")

# Garantir que o número de arquivos seja igual
if len(input_files) == 0 or len(target_files) == 0:
    raise ValueError("Nenhum arquivo .wav encontrado nos diretórios de entrada ou saída.")
if len(input_files) != len(target_files):
    raise ValueError("Número de arquivos de entrada e saída não coincide.")

# Dividir em treino e validação (80% treino, 20% validação)
train_inputs, val_inputs, train_targets, val_targets = train_test_split(
    input_files, target_files, test_size=0.2, random_state=42
)

# Mover os arquivos de treino
for file in train_inputs:
    shutil.copy(file, os.path.join(train_input_dir, os.path.basename(file)))
for file in train_targets:
    shutil.copy(file, os.path.join(train_target_dir, os.path.basename(file)))

# Mover os arquivos de validação
for file in val_inputs:
    shutil.copy(file, os.path.join(val_input_dir, os.path.basename(file)))
for file in val_targets:
    shutil.copy(file, os.path.join(val_target_dir, os.path.basename(file)))

print("Divisão completa!")
```
