```python
import torch
import librosa
import soundfile as sf
from pathlib import Path
```

```python
# Load TorchScript model
model_path = "../models/chorus_v4_streaming.ts"
model_name = Path(model_path).stem
model = torch.jit.load(model_path)
```

```python
# Load audio file 
audio_path = "../raw_data/calming-electric-guitar_70bpm_F_major.wav"
file_name = Path(audio_path).name
audio, sr = librosa.load(audio_path, sr=None)
```

```python
# Convert the audio to a tensor and add the batch dimension
input_audio = torch.tensor(audio).unsqueeze(0)  # Add batch dimension
input_audio = input_audio.unsqueeze(0)  # Add channel dimension (mono)
```

```python
with torch.no_grad():  # Disable autograd to speed up inference
    output_audio = model(input_audio)
```

```python
# Remove dimensions of size 1 from the output tensor (batch and channel dimensions)
output_audio_np = output_audio.squeeze().numpy()
```

```python
sf.write(f"../transformed_data/{model_name}_{file_name}", output_audio_np, sr)
```
